```
import pandas as pd
```

```
link = pd.read_csv("Dataset/link.csv")
movie = pd.read_csv("Dataset/movie.csv")
rating = pd.read_csv("Dataset/rating.csv")
tag = pd.read_csv("Dataset/tag.csv")
genome_score = pd.read_csv("Dataset/genome_scores.csv")
genome_tag = pd.read_csv("Dataset/genome_tags.csv")
```

```
movie["title"].unique()
```

```
grouped_tag = tag.groupby("movieId")["tag"].apply(lambda x: " ".join([str(i) for i in set(x) if isinstance(i, str)]))
```

```
grouped_tag.shape
```

```
genome_tag.shape
```

```
genome_tag.shape
```

```
movie.drop_duplicates()
movie.dropna()
```

```
grouped_rating = rating.groupby("movieId")["rating"].mean().reset_index()
```

```
Dataset = pd.merge(movie, grouped_tag, on="movieId", how="inner")
```

```
Dataset.shape
```

```
Dataset = pd.merge(Dataset, link, on="movieId", how="inner")
```

```
Dataset.shape
```

```
df = pd.merge(Dataset, grouped_rating, on="movieId", how="inner")
df.shape
```
