# Evaluation Metrics
Precision, Recall, and NDCG scores

```
from sklearn.metrics import ndcg_score

# Evaluation Functions
def precision_at_k(relevant_items, recommended_items, k=10):
    """Calculate Precision@K
    Args:
        relevant_items: list of relevant movie titles
        recommended_items: list of recommended movie titles
        k: number of top recommendations to consider
    Returns:
        precision score (0-1)
    """
    rec_k = recommended_items[:k]
    relevant_in_k = len(set(rec_k) & set(relevant_items))
    return relevant_in_k / k if k > 0 else 0

def recall_at_k(relevant_items, recommended_items, k=10):
    """Calculate Recall@K
    Args:
        relevant_items: list of relevant movie titles
        recommended_items: list of recommended movie titles
        k: number of top recommendations to consider
    Returns:
        recall score (0-1)
    """
    rec_k = recommended_items[:k]
    relevant_in_k = len(set(rec_k) & set(relevant_items))
    return relevant_in_k / len(relevant_items) if len(relevant_items) > 0 else 0

def evaluate_recommendations(test_user_likes, recommended_movies, k=10):
    """Calculate all evaluation metrics
    Args:
        test_user_likes: list of movie titles user actually liked (ground truth)
        recommended_movies: list of recommended movie titles (ranked)
        k: number of top recommendations to evaluate
    Returns:
        dict with precision, recall, and ndcg scores
    """
    relevant_set = set(test_user_likes)
    recommended_k = recommended_movies[:k]
    
    # Precision & Recall
    precision = len(set(recommended_k) & relevant_set) / k
    recall = len(set(recommended_k) & relevant_set) / len(relevant_set) if len(relevant_set) > 0 else 0
    
    # NDCG (1 if in relevant_set, 0 otherwise)
    y_true = [1 if movie in relevant_set else 0 for movie in recommended_k]
    y_score = list(range(len(recommended_k), 0, -1))  # decreasing scores
    
    ndcg = ndcg_score([y_true], [y_score], k=k)
    
    return {
        'precision@k': precision,
        'recall@k': recall,
        'ndcg@k': ndcg
    }

print('Evaluation functions loaded!')
movie_name = "iron man"
closest_match = difflib.get_close_matches(movie_name, movie_list, cutoff=0.3)
movie_index = df[df.title == closest_match[0]].index[0]
```

```
# Example: Test with ground truth data
# Use actual movie titles from the dataset (with year)
test_user_likes = ["Iron Man (2008)", "Avengers, The (2012)", "Captain America: The First Avenger (2011)"]  # Real titles

# Get recommendations for a movie
movie_name = "iron man"
closest_match = difflib.get_close_matches(movie_name, movie_list, cutoff=0.3)
movie_index = df[df.title == closest_match[0]].index[0]

recoms_full = sorted(enumerate(feature_similarity[movie_index]), key=lambda x: x[1], reverse=True)
recommended_movies = [df.iloc[recoms_full[i][0]]['title'] for i in range(len(recoms_full))]

print(f'Query movie: {closest_match[0]}')
print(f'Ground truth (liked movies): {test_user_likes}')
print(f'Top 10 recommendations: {recommended_movies[:10]}')
print()

# Calculate scores
scores = evaluate_recommendations(test_user_likes, recommended_movies, k=10)
print(f'Evaluation Metrics for Top 10 Recommendations:')
print(f'Precision@10: {scores["precision@k"]:.3f}')
print(f'Recall@10: {scores["recall@k"]:.3f}')
print(f'NDCG@10: {scores["ndcg@k"]:.3f}')
```

```
def rmse(y_true, y_pred):
    """Calculate RMSE (Root Mean Squared Error)
    Args:
        y_true: actual ratings
        y_pred: predicted ratings/similarity scores
    Returns:
        RMSE score
    """
    return np.sqrt(np.mean((np.array(y_true) - np.array(y_pred)) ** 2))

print('RMSE function loaded!')
```

```
# Calculate RMSE for recommendations
# Compare similarity scores to actual ratings

# Get similarity scores for top 10 recommendations
predicted_scores = [recoms_full[i][1] for i in range(10)]

# Get actual ratings of recommended movies
actual_ratings = [df.iloc[recoms_full[i][0]]['rating'] for i in range(10)]

# Normalize predicted scores (similarity is 0-1, ratings vary)
predicted_ratings = [score * 5 for score in predicted_scores]  # scale to 0-5

rmse_score = rmse(actual_ratings, predicted_ratings)

print(f'\nAdditional Metrics:')
print(f'RMSE@10: {rmse_score:.3f}')
print(f'\nActual ratings: {[f"{r:.2f}" for r in actual_ratings]}')
print(f'Predicted ratings: {[f"{r:.2f}" for r in predicted_ratings]}')
```

```
import numpy as np
from sklearn.metrics import ndcg_score

def precision_at_k(relevant, recommended, k=10):
    rec_k = set(recommended[:k])
    rel_set = set(relevant)
    return len(rec_k & rel_set) / k if k > 0 else 0

def recall_at_k(relevant, recommended, k=10):
    rec_k = set(recommended[:k])
    rel_set = set(relevant)
    return len(rec_k & rel_set) / len(rel_set) if len(rel_set) > 0 else 0

movie_query = 'iron man'
ground_truth = ['Iron Man (2008)', 'Avengers, The (2012)', 'Captain America: The First Avenger (2011)', 'Thor (2011)']

closest = difflib.get_close_matches(movie_query, movie_list, cutoff=0.3)
if closest:
    movie_idx = df[df.title == closest[0]].index[0]
    movie_id = df.loc[movie_idx, 'movieId']
    
    print(f'Query: {closest[0]}')
    print(f'Ground truth: {ground_truth}')
    print()
    
    print('='*50)
    print('1. CONTENT-BASED (TFIDF)')
    print('='*50)
    sim1 = feature_similarity[movie_idx]
    top1 = sorted(enumerate(sim1), key=lambda x: x[1], reverse=True)[1:11]
    recs1 = [df.iloc[idx]['title'] for idx, _ in top1]
    p1 = precision_at_k(ground_truth, recs1, 10)
    r1 = recall_at_k(ground_truth, recs1, 10)
    print(f'Precision@10: {p1:.4f}')
    print(f'Recall@10: {r1:.4f}')
    print(f'Top 5: {recs1[:5]}')
    print()
    
    print('='*50)
    print('2. USER-BASED COLLABORATIVE FILTERING')
    print('='*50)
    user_idx = 0
    sim2 = user_similarity[user_idx]
    similar_users = sorted(enumerate(sim2), key=lambda x: x[1], reverse=True)[1:6]
    
    recs_dict = {}
    for u_idx, u_sim in similar_users:
        for m_idx, rating in enumerate(user_movie_matrix.iloc[u_idx]):
            if rating > 0:
                recs_dict[m_idx] = recs_dict.get(m_idx, 0) + u_sim * rating
    
    top2_idx = sorted(recs_dict.items(), key=lambda x: x[1], reverse=True)[:10]
    recs2 = [df[df['movieId'] == user_movie_matrix.columns[idx]].iloc[0]['title'] for idx, _ in top2_idx]
    p2 = precision_at_k(ground_truth, recs2, 10)
    r2 = recall_at_k(ground_truth, recs2, 10)
    print(f'Precision@10: {p2:.4f}')
    print(f'Recall@10: {r2:.4f}')
    print(f'Top 5: {recs2[:5]}')
    print()
    
    print('='*50)
    print('3. ITEM-BASED COLLABORATIVE FILTERING')
    print('='*50)
    movie_matrix_idx = list(user_movie_matrix.columns).index(movie_id)
    sim3 = movie_similarity[movie_matrix_idx]
    top3 = sorted(enumerate(sim3), key=lambda x: x[1], reverse=True)[1:11]
    recs3 = [df[df['movieId'] == user_movie_matrix.columns[idx]].iloc[0]['title'] for idx, _ in top3]
    p3 = precision_at_k(ground_truth, recs3, 10)
    r3 = recall_at_k(ground_truth, recs3, 10)
    print(f'Precision@10: {p3:.4f}')
    print(f'Recall@10: {r3:.4f}')
    print(f'Top 5: {recs3[:5]}')
    print()
    
    print('='*50)
    print('SUMMARY')
    print('='*50)
    models = [('Content-Based', p1, r1), ('User-Based', p2, r2), ('Item-Based', p3, r3)]
    for name, p, r in models:
        print(f'{name:20s} | P@10: {p:.4f} | R@10: {r:.4f}')
    
    winner = max(models, key=lambda x: (x[1] + x[2]) / 2)
    print()
    print(f'🏆 Winner: {winner[0]}')
```
